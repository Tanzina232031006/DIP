# 2.2 Gamma Correction

## Objective
To adjust the brightness of an image using Gamma Correction.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# 1. Read Image
# ==========================================

image = cv2.imread('/content/image.png')

# Check if image is loaded
if image is None:
    print("Error: Image not found!")
else:
    # Convert BGR to RGB
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # ==========================================
    # 2. Gamma Correction
    # ==========================================

    gamma = 2.0

    # Create lookup table
    table = np.array([
        ((i / 255.0) ** (1.0 / gamma)) * 255
        for i in np.arange(256)
    ]).astype("uint8")

    # Apply Gamma Correction
    gamma_corrected = cv2.LUT(image_rgb, table)

    # ==========================================
    # 3. Display Results
    # ==========================================

    plt.figure(figsize=(12, 5))

    # Original Image
    plt.subplot(1, 2, 1)
    plt.imshow(image_rgb)
    plt.title("Original Image")
    plt.axis("off")

    # Gamma Corrected Image
    plt.subplot(1, 2, 2)
    plt.imshow(gamma_corrected)
    plt.title("Gamma Correction (γ = 2.0)")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
```
