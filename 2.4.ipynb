# 2.4 Bit Plane Slicing

## Objective
To extract and display the individual bit planes of a grayscale image.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# 1. Read Image
# ==========================================

image = cv2.imread('/content/image.png')

# Check if image is loaded
if image is None:
    print("Error: Image not found!")

else:
    # ==========================================
    # 2. Convert to Grayscale
    # ==========================================

    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # ==========================================
    # 3. Display Original + Bit Planes
    # ==========================================

    plt.figure(figsize=(12, 10))

    # Original Image
    plt.subplot(3, 3, 1)
    plt.imshow(gray, cmap='gray')
    plt.title("Original Image")
    plt.axis("off")

    # ==========================================
    # 4. Extract Bit Planes
    # ==========================================

    for bit in range(8):

        # Extract bit plane
        bit_plane = (gray >> bit) & 1

        # Convert to 0-255 for display
        bit_plane = bit_plane * 255

        # Display
        plt.subplot(3, 3, bit + 2)
        plt.imshow(bit_plane, cmap='gray')
        plt.title(f"Bit Plane {bit}")
        plt.axis("off")

    # ==========================================
    # 5. Overall Title
    # ==========================================

    plt.suptitle("2.4 Bit Plane Slicing", fontsize=16)
    plt.tight_layout()
    plt.show()
```
