```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# STEP 1: LOAD IMAGE
filename = "/content/image.png"

image = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)

if image is None:
    raise ValueError("Image not found! Check the file name.")

image = cv2.resize(image, (256, 256))

# STEP 2: APPLY HISTOGRAM EQUALIZATION
equalized = cv2.equalizeHist(image)

# STEP 3: GENERATE HISTOGRAMS
original_hist = cv2.calcHist(
    [image], [0], None, [256], [0, 256]
).flatten()

equalized_hist = cv2.calcHist(
    [equalized], [0], None, [256], [0, 256]
).flatten()

# STEP 4: DISPLAY RESULTS
plt.figure(figsize=(14, 9))

# Original Image
plt.subplot(2, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image - Low Contrast")
plt.axis("off")

# Equalized Image
plt.subplot(2, 2, 2)
plt.imshow(equalized, cmap="gray")
plt.title("Equalized Image - Enhanced Contrast")
plt.axis("off")

# Original Histogram
plt.subplot(2, 2, 3)
plt.plot(original_hist, color="blue")
plt.title("Original Histogram")
plt.xlabel("Gray Level")
plt.ylabel("Frequency")
plt.xlim([0, 256])
plt.grid(alpha=0.2)

# Equalized Histogram
plt.subplot(2, 2, 4)
plt.plot(equalized_hist, color="red")
plt.title("Equalized Histogram")
plt.xlabel("Gray Level")
plt.ylabel("Frequency")
plt.xlim([0, 256])
plt.grid(alpha=0.2)

plt.suptitle(
    "Assignment 3.4: Histogram Equalization",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()

# STEP 5: SAVE OUTPUT
plt.savefig(
    "Assignment_3_4_Histogram_Equalization.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# STEP 6: COMPLETION MESSAGE
print("Histogram Equalization Completed!")
print("Original Image Size:", image.shape)
print("Equalized Image Size:", equalized.shape)
print("Formula: sk = (L - 1) x CDF(rk)")
print("For 8-bit image, L = 256")
```
