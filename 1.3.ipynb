# 1.3 Spatial Interpolation

## Objective
To resize an image using Nearest Neighbor, Bilinear, and Bicubic interpolation methods and compare their results.

```python
import cv2
import matplotlib.pyplot as plt

# ==========================================
# 1. Read Image
# ==========================================

image = cv2.imread('/content/image.png')

# Check if image is loaded
if image is None:
    print("Error: Image not found!")
else:
    # Convert BGR to RGB
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # Get original dimensions
    height, width = image_rgb.shape[:2]

    # ==========================================
    # 2. New Image Size
    # ==========================================

    new_width = width * 2
    new_height = height * 2

    # ==========================================
    # 3. Nearest Neighbor
    # ==========================================

    nearest = cv2.resize(
        image_rgb,
        (new_width, new_height),
        interpolation=cv2.INTER_NEAREST
    )

    # ==========================================
    # 4. Bilinear
    # ==========================================

    bilinear = cv2.resize(
        image_rgb,
        (new_width, new_height),
        interpolation=cv2.INTER_LINEAR
    )

    # ==========================================
    # 5. Bicubic
    # ==========================================

    bicubic = cv2.resize(
        image_rgb,
        (new_width, new_height),
        interpolation=cv2.INTER_CUBIC
    )

    # ==========================================
    # 6. Display Results
    # ==========================================

    plt.figure(figsize=(16, 10))

    # Original
    plt.subplot(2, 2, 1)
    plt.imshow(image_rgb)
    plt.title("Original Image")
    plt.axis("off")

    # Nearest Neighbor
    plt.subplot(2, 2, 2)
    plt.imshow(nearest)
    plt.title("Nearest Neighbor")
    plt.axis("off")

    # Bilinear
    plt.subplot(2, 2, 3)
    plt.imshow(bilinear)
    plt.title("Bilinear")
    plt.axis("off")

    # Bicubic
    plt.subplot(2, 2, 4)
    plt.imshow(bicubic)
    plt.title("Bicubic")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
```
