# 2.1 Linear Transformation

## Objective
To improve the contrast of an image using linear transformation.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# 1. Read Image
# ==========================================

image = cv2.imread('/content/image.png')

# Check if image is loaded
if image is None:
    print("Error: Image not found!")
else:
    # Convert BGR to RGB
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # ==========================================
    # 2. Linear Transformation
    # ==========================================

    # Contrast and brightness parameters
    alpha = 1.5
    beta = 20

    # Apply linear transformation
    transformed = cv2.convertScaleAbs(
        image_rgb,
        alpha=alpha,
        beta=beta
    )

    # ==========================================
    # 3. Display Results
    # ==========================================

    plt.figure(figsize=(12, 5))

    # Original Image
    plt.subplot(1, 2, 1)
    plt.imshow(image_rgb)
    plt.title("Original Image")
    plt.axis("off")

    # Linear Transformed Image
    plt.subplot(1, 2, 2)
    plt.imshow(transformed)
    plt.title("Linear Transformation")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
```
