# 2.1 Linear Transformation

## Objective
To improve the contrast of an image using linear transformation.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# 1. Read Image
# ==========================================

image = cv2.imread('/content/image.png')

# Check if image is loaded
if image is None:
    print("Error: Image not found!")

else:
    # Convert BGR to RGB
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # ==========================================
    # 2. Contrast Transformation
    # Formula: s = a*r
    # ==========================================

    a = 1.8

    contrast = a * image_rgb.astype(np.float32)

    # Keep values between 0 and 255
    contrast = np.clip(contrast, 0, 255)

    # Convert to uint8
    contrast = contrast.astype(np.uint8)

    # ==========================================
    # 3. Brightness Transformation
    # Formula: s = r + b
    # ==========================================

    b = 50

    brightness = image_rgb.astype(np.float32) + b

    # Keep values between 0 and 255
    brightness = np.clip(brightness, 0, 255)

    # Convert to uint8
    brightness = brightness.astype(np.uint8)

    # ==========================================
    # 4. Linear Transformation
    # Formula: s = a*r + b
    # ==========================================

    a = 1.5
    b = 20

    linear = a * image_rgb.astype(np.float32) + b

    # Keep values between 0 and 255
    linear = np.clip(linear, 0, 255)

    # Convert to uint8
    linear = linear.astype(np.uint8)

    # ==========================================
    # 5. Display Results
    # ==========================================

    plt.figure(figsize=(15, 10))

    # Original
    plt.subplot(2, 2, 1)
    plt.imshow(image_rgb)
    plt.title("Original Image")
    plt.axis("off")

    # Contrast
    plt.subplot(2, 2, 2)
    plt.imshow(contrast)
    plt.title("Contrast (a = 1.8)")
    plt.axis("off")

    # Brightness
    plt.subplot(2, 2, 3)
    plt.imshow(brightness)
    plt.title("Brightness (b = 50)")
    plt.axis("off")

    # Linear Transformation
    plt.subplot(2, 2, 4)
    plt.imshow(linear)
    plt.title("Linear Transformation (a = 1.5, b = 20)")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
```
